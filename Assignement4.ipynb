## **Problem 1: Mobile Recharge Validation System**

**Problem Statement**

Create a function to validate a mobile recharge request.

**Rules:**

- Recharge amount must be ≥ ₹50


- Valid plans: 199, 299, 399, 599


- User should be allowed to retry if invalid


**Real-Time Use**

- Telecom recharge apps


- Payment gateways


**Hint**

- Use function for validation


- Use while loop for retry

````python
# Mobile Recharge Validation System

def validate_recharge(amount, plan):
    """
    This function validates the recharge amount and selected plan.
    Returns True if valid, otherwise False.
    """
    
    valid_plans = [199, 299, 399, 599]
    
    # Checking minimum amount
    if amount < 50:
        print("Recharge amount must be at least ₹50.")
        return False
    
    if plan not in valid_plans:
        print("Invalid plan selected. Available plans:", valid_plans)
        return False
    
    return True


# if not satisfied

while True:
    try:
        amount = float(input("Enter recharge amount: ₹"))
        plan = int(input("Enter recharge plan (199, 299, 399, 599): "))
        
        if validate_recharge(amount, plan):
            print("Recharge Successful ")
            break  
        else:
            print("Please try again.\n")
            
    except ValueError:
        print("Invalid input! Please enter numeric values.\n")
````

## **Problem 2: Inventory Reorder Alert System**

**Problem Statement**

Given a dictionary of products and stock quantities:
Print Reorder Alert if stock < 15


Print Stock OK otherwise


**Real-Time Use**

- Warehouse systems


- Retail management


**Hint**

- Use function


- Use for loop on dictionary




````python
# Inventory Reorder Alert System

def check_inventory(stock_dict):
    """
    This function checks stock levels of products.
    Prints 'Reorder Alert' if stock < 15
    Otherwise prints 'Stock OK'
    """
    
    for product, quantity in stock_dict.items():
        if quantity < 15:
            print(f"{product}: Reorder Alert ⚠ (Stock = {quantity})")
        else:
            print(f"{product}: Stock OK  (Stock = {quantity})")


# sample test case
inventory = {
    "Rice": 20,
    "Wheat": 10,
    "Sugar": 5,
    "Oil": 25,
    "Salt": 14
}

# calling the function
check_inventory(inventory)
````

# Problem 4: Cab Fare Estimator with Retry Option

##  Problem Statement

Create a cab fare calculator with the following conditions:

- **Base Fare** = ₹50  
- **Charge per kilometer** = ₹12  
- **Peak hour** → 25% extra charge  
- Allow user to **retry** using a `while` loop  

---

## Real-Time Use

- Ride-hailing apps  
- Transport systems  

---


```


````python

def calculate_fare(distance, is_peak):
    base_fare = 50
    per_km = 12
    
    fare = base_fare + (distance * per_km)
    
    if is_peak == "yes":
        fare = fare + (fare * 0.25)   # Add 25% extra
    
    return fare


# Retry option
while True:
    distance = float(input("Enter distance (km): "))
    
    if distance <= 0:
        print("Invalid distance. Try again.\n")
        continue
    
    peak = input("Is it peak hour? (yes/no): ").lower()
    
    total = calculate_fare(distance, peak)
    
    print("Total Fare = ₹", total)
    break
````

#  Problem 5: Employee Attendance Eligibility Checker

##  Problem Statement

Create a function that:

- Accepts an attendance list containing `"P"` (Present) and `"A"` (Absent)
- Calculates attendance percentage
- Returns **"Eligible"** if attendance ≥ 75%  
- Returns **"Not Eligible"** otherwise  

---

##  Real-Time Use

- HR payroll systems  
- Internship tracking systems  

---

##  Hint

- Use a loop to count present days  
- Use a return statement  



````python

def check_eligibility(attendance_list):
    total_days = len(attendance_list)
    present_days = 0

    # Count present days
    for day in attendance_list:
        if day == "P":
            present_days += 1

    # Calculate percentage
    percentage = (present_days / total_days) * 100

    # Return eligibility result
    if percentage >= 75:
        return "Eligible"
    else:
        return "Not Eligible"


# Example usage
attendance = ["P", "A", "P", "P", "A", "P", "P", "P"]

result = check_eligibility(attendance)
print("Attendance Percentage:", (attendance.count("P") / len(attendance)) * 100)
print("Eligibility Status:", result)
````

#  Problem 6: Password Strength Checker

##  Problem Statement

Create a function to check password strength.

### Rules:
- Minimum length = 8  
- Must contain at least one digit  
- Must contain at least one special character (@ # $)  

---

##  Real-Time Use

- Login systems  
- Security validation  

---

##  Hint

- Use a loop over the string  
- Use conditions inside the function  

---



````python

def check_password(password):
    
    # Rule 1: Minimum length
    if len(password) < 8:
        return "Weak Password (Minimum 8 characters required)"
    
    has_digit = False
    has_special = False
    special_chars = "@#$"
    
    # Loop through password characters
    for char in password:
        if char.isdigit():
            has_digit = True
        if char in special_chars:
            has_special = True
    
    # Check final conditions
    if has_digit and has_special:
        return "Strong Password "
    else:
        return "Weak Password (Must include at least one digit and one special character)"


# Taking user input
password = input("Enter your password: ")
result = check_password(password)

print(result)
````
